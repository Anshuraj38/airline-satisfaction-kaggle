## Data Quality Analysis

```python
missing = train.isnull().sum()

missing_percentage = train.isnull().mean() * 100

missing_report = pd.DataFrame({
    'Missing': missing,
    'Missing Percentage': missing_percentage
})

missing_report = missing_report.sort_values(
    by='Missing Percentage',
    ascending=False
    )

print(missing_report.head(5))
```

```python
# Vizualize the Distribution 

plt.figure(figsize=(7,5))

sns.countplot(
    data=train,
    x='satisfaction'
)

plt.title('Distribution of Pasenger Satisfation')
plt.xlabel('Satisfaction')
plt.ylabel('Number of Passenger')
```

```python
# Let's understand the categorical variables

categorical_columns = train.select_dtypes(
    include=['object','category']
).columns

for col in categorical_columns:
    
    print(f'\n{train[col].value_counts()}')
```
